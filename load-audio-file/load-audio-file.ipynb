```python
import torch
import soundfile as sf
```

```python
audio, sample_rate = sf.read("/Users/mehrj/Documents/Music/Kits/MetroBoomin Drum Kit/Loops/Metro_loopC#m (1).wav")
print("audio shape:", audio.shape)
print("sample rate:", sample_rate)
print("duration:", audio.shape[0]/sample_rate)
```

```python
waveform = torch.tensor(audio, dtype=torch.float32)
print("shape", waveform.shape)
print("data type", waveform.dtype)
print("first five samples", waveform[:5])
```

```python
import matplotlib.pyplot as plt

time = torch.arange(waveform.shape[0]) / sample_rate

n = int(1 * sample_rate)
plt.plot(time[:n].numpy(), waveform[:n].numpy())
plt.xlabel("time")
plt.ylabel("amplitude")
plt.title("audio waveform")
```

```python
if waveform.ndim == 2:
    mono = waveform.mean(dim=1)
else:
    mono = waveform

plt.specgram(mono.numpy(), NFFT = 2000, Fs = sample_rate, noverlap=1)
plt.xlabel("Time (seconds)")
plt.ylabel("Frequency (Hz)")
plt.colorbar(label='power(db)')
```

```python
from IPython.display import Audio, display

display(Audio(mono.numpy(), rate=sample_rate))
```
